```python
#Filling Missing Values
```

```python
products[products["Cost_Price_INR"].isnull()]
```

```python
Category_Ratio = products.groupby("Category")["Cost_Price_INR"].transform("median")

products["Cost_Price_INR"] = products["Cost_Price_INR"].fillna(products["Base_Price_INR"]*Category_Ratio)
```

```python
#Filtering Negative Values
```

```python
products[["Cost_Price_INR","Base_Price_INR"]].describe()
```

```python
#Storing Negative Values
Negative_Products = products.loc[products["Base_Price_INR"] < 0, "Product_ID"] 

sales[sales["Product_ID"].isin(Negative_Products)].shape
```

```python
#Filtering
products = products[products["Base_Price_INR"] >=0]
```

```python
Missing_Quantity = sales[sales["Quantity"].isnull()].copy()
```

```python
Missing_Quantity["Calculated_Qty"] = (Missing_Quantity["Sales_Amount_INR"]/Missing_Quantity["Unit_Price_INR"])
```

```python
#Validating Sales
```

```python
Negative_Mask = sales["Quantity"] < 0
```

```python
sales.loc[Negative_Mask,"Quantity"] = (sales.loc[Negative_Mask,"Sales_Amount_INR"]/sales.loc[Negative_Mask,"Unit_Price_INR"]).round()
```

```python
sales["Quantity"] = sales["Quantity"].fillna(sales["Sales_Amount_INR"]/sales["Unit_Price_INR"]).round()
```

```python
sales[sales["Quantity"].isnull()].shape
```

```python
#Standardizing Text Values
sales[["Sales_Channel","Payment_Method","Order_Status"]] = sales[["Sales_Channel","Payment_Method","Order_Status"]].apply(lambda x : x.str.title())
```

```python
sales["Discount"].value_counts(dropna=False).sort_index()
```

```python
sales["Order_Status"] = sales["Order_Status"].fillna("Unknown")
```

```python
sales["Order_Status"] = sales["Order_Status"].str.replace("Nan","Unknown")
```
